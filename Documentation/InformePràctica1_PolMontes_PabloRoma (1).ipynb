# Projectes d'Enginyeria: Informe de Pràctica1

### Introducció

Aquesta pràctica de laboratori consisteix a familiaritzar-se amb tot l’entorn tecnològic que s’utilitzarà a l’assignatura, que inclou entre d’altres la plataforma GitHub, el PlatformIO i l’ús de l'Arduino.


### Objectiu

L’objectiu d’aquesta pràctica és prendre contacte amb el sistema Arduino i entendre els seus fonaments per a la futura implementació d’un sistema similar per al projecte en grup de l’assignatura.

### Mètode

Abans de començar a fer servir Arduino, s’ha començat per familiaritzar-se amb la plataforma GitHub, aprenent a gestionar-hi documents i fer-la servir com a base per a projectes compartits.

Per aquesta raó, vam començar creant-nos un compte a la plataforma, definint un directori, fent un Fork del repositori de projectes del perfil de Manel Puig i establint els col·laboradors. A continuació, vam aprendre a crear un clon de les carpetes a l’ordinador utilitzat durant la sessió per a poder treballar directament al repositori del director del grup des de Visual Studio Code. En aquest punt vam fer alguns petits canvis de prova, els vam sincronitzar amb el GitHub del director, afegint-hi un comentari de “Canvis de prova”, i vam comprovar que tot havia succeït correctament.

Una vegada familiaritzats amb el procés de treball en projectes partint de GitHub, vam crear un projecte nou amb l’eina de PlatformIO, dins del mateix Visual Studio Code. En aquest es va utilitzar un codi ja proporcionat pel guió de la pràctica al mateix PlatformIO, i un codi de C++ a la carpeta de src/main.cpp. Aquest codi consistia a fer que el LED d’una placa d’ESP32 s'encengués i s’apagués alternament cada segon, i a la pantalla de l’ordinador es mostrés un missatge cada vegada que l’estat del llum canviés.

Així doncs, vam connectar la placa que es volia programar per mitjà d’un cable USB a l’ordinador i hi vam compilar el codi. Com a resultat, el llum va funcionar com s’esperava.

A continuació, vam introduir-nos a treballar amb un sensor Inertial Mass Unit (IMU), que inclou un giroscopi per a mesurar velocitats angulars, acceleròmetre per a mesurar acceleracions lineals, compàs per a referenciar l'orientació i un DMP microcontroller per a tenir una adequada orientació 3D. Aquest sensor va ser connectat amb una bateria.

![muntatge](Images/muntatge.jpg)

Una vegada connectats via Wi-Fi correctament al microprocessador amb el sensor, propi de cada grup, vam pujar-li el codi proporcionat Endowrist_IMU mitjançant el mencionat PlatformIO.

A més, va ser necessari obrir el fitxer 3D_Orientation.rdk al programa roboDK. Aquest programa ens va permetre simular que el nostre sensor IMU fos un avió o una agulla de quiròfan. Mitjançant la correcta orientació del sensor, vam observar com la simulació al programa dels diferents objectes responia correctament als canvis d’orientació que fèiem en el dispositiu.


Finalment, vam guardar el nostre projecte i vam aconseguir pujar els canvis en el GitHub de manera satisfactòria.


### Conclusions

En aquesta pràctica hem aconseguit assolir els objectius que ens havíem proposat inicialment. Primer, hem aconseguit familiaritzar-nos amb la plataforma GitHub, on hem guanyat experiència per a gestionar correctament projectes compartits.

A més, hem après a utilitzar l’entorn de PlatformIO dins de Visual Studio Code, fonamental per a desar correctament canvis en el codi i fer les primeres proves de diferents funcions del programa.

També ha estat essencial per al futur del projecte de l’assignatura guanyar confiança amb el sistema Arduino mitjançant diferents proves. A més, vam aconseguir els resultats proposats utilitzant la placa ESP32 i el LED de manera satisfactòria.

Finalment, vam aconseguir entendre el funcionament del sensor IMU per a obtenir informació sobre l’orientació del dispositiu en l’espai. A més, gràcies al programa RoboDK, vam comprovar com aquests canvis d’orientació es reflecteixen correctament en les simulacions tridimensionals del programa.

Com a resum, hem aconseguit adquirir coneixements bàsics amb diferents eines de programació i l’ús de sensors, serà fonamental per al desenvolupament del projecte final de l'assignatura.
